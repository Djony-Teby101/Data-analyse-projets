```python
import pandas as pd
import numpy as np

np.random.seed(52)
dates=pd.date_range(start='2020-01-01',end='2023-12-31', freq='MS')
ventes_data=[]

for magasin in range(1,6):
    for date in dates:
        mois=date.month
        annee=date.year

        facteur_saisonnier= 1 + 0.3 * np.sin((mois-6)*np.pi/6)**2
        if mois ==12:
            facteur_saisonnier *=1.4

            # Tendance: croissance annuelle de 5%
            facteur_tendance=1 + 0.05 * (annee-2020)

            # Base par magasin en ordre decroissant.
            base_magasin=[12000,9000, 10500,7000,9500][magasin-1]

            #Bruit aleatoire.
            bruit=np.random.normal(1, 0.08)

            # Calcul du chiffre d'affaire
            ca=base_magasin*facteur_saisonnier*facteur_tendance*bruit
            ca=max(0, ca) # pas de chiffre negatif.

            ventes_data.append({
                'id_magasin': magasin,
                'date':date.strftime('%Y-%m-%d'),
                'chiffre_affaires':round(ca, 2),
                'region_ville':'Libreville'
            })

df_ventes=pd.DataFrame(ventes_data)
```

```python
# Recuperer le fichier pour l'analyse.
df_ventes.to_csv('ventes.csv', index=False)
```
